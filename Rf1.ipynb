```python
x = df.iloc[:,:-1]
x
```

```python
y=df.iloc[:,-1:]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.20,random_state=0,stratify=y)
```

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(n_estimators=20,criterion='gini',max_depth=5,max_features=5)

rfc.fit(x_train,y_train)
```

```python
y_pred = rfc.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
test_acc = classification_report(y_test,y_pred)
print(test_acc)
```

```python
y_train_pred = rfc.predict(x_train)
```

```python
train_acc = accuracy_score(y_train_pred,y_train)
train_acc
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(rfc,x,y,cv=3)

print('CV scores :',cv_scores)
print('Mean CV Accuracy :',cv_scores.mean())
print('CV std accuracy :',cv_scores.std())
```

```python
# good model coz high mean and low std 
```

```python
from sklearn.model_selection import learning_curve

train_sizes,train_scores,test_scores = learning_curve(rfc,x,y,cv=3)
```

```python
import numpy as np
import matplotlib.pyplot as plt


train_mean = train_scores.mean(axis=1)
test_mean = test_scores.mean(axis=1)

plt.plot(train_sizes,train_mean,marker='o',label='Training Accuracy')
plt.plot(train_sizes,test_mean,marker='o',label='Validation Accuracy')
plt.xlabel('Training set size')
plt.xlabel('Accuracy')
plt.title('Learning curve ')
plt.legend()
plt.show()
```
