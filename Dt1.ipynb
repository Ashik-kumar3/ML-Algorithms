```python
df['Insulin'].unique()
```

```python
df['Insulin']=df['Insulin'].replace({'Zero':0})
df['Insulin']=df['Insulin'].astype(int)
```

```python
df.info()
```

```python
df['Outcome'].unique()
```

```python
df['Outcome']=df['Outcome'].replace({'N':0,'Y':1})
df['Outcome']=df['Outcome'].astype(int)
```

```python
df.info()
```

```python
df.corr()['Outcome']
```

```python
df.drop(columns=['Pregnancies','SkinThickness','Insulin'],inplace=True)
```

```python
df.head()
```

```python
for i in df.columns:
    sns.boxplot(df[i])
    plt.show()
```

```python
remove = ['Glucose','Age']
for i in remove:
    qrt1 = df[i].quantile(.25)
    qrt3 = df[i].quantile(.75)
    iqr = qrt3-qrt1
    uw = min((qrt3+1.5*iqr),(df[i].max()))
    lw = max((qrt1-1.5*iqr),(df[i].min()))
    df = df[(df[i]>lw) & (df[i]<uw)]
    plt.boxplot(df[i])
    plt.title(i)
    plt.show()
```

```python
df=df[df['BMI']!=0]
df=df[df['BloodPressure']!=0]
```

```python
for i in df.columns:
    sns.boxplot(df[i])
    plt.show()
```

```python
df.head()
```

```python
df.shape
```

```python
df['Outcome']
```

```python
x=df.iloc[:,:5]
y=df['Outcome']
```

```python
from sklearn.feature_selection import SelectKBest
bestfeatures = SelectKBest(k='all')
fit = bestfeatures.fit(x,y)
print(fit.scores_)
```

```python
fit
```

```python
dfscore = pd.DataFrame(fit.scores_)
```

```python
dfscore
```

```python
dfcolumns = pd.DataFrame(x.columns)
fs = pd.concat([dfcolumns,dfscore],axis=1)
fs.column = ['spec','score']
```

```python
fs
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=0,stratify=y)
```

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE()
```

```python
x_res,y_res = sm.fit_resample(x_train,y_train)
```

```python
y_res.value_counts()
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(max_depth=5,criterion='entropy',splitter='best',min_samples_split=60)
dt.fit(x_train,y_train)
y_pred=dt.predict(x_test)
y_pred
```

```python
help(dt)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_pred,y_test))
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_pred,y_test)
acc
```

```python
y_train_pred=dt.predict(x_train)
y_train_pred
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_train_pred,y_train))
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_train_pred,y_train)
acc
```

```python
if accuracy_score(y_train,y_train_pred)<.60:
    print("overfit")

elif accuracy_score(y_train,y_train_pred) > accuracy_score(y_test,y_pred)+.11:
    print("underfit")

else:
    print("good fit")
```
