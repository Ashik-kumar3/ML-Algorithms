```python
df = df[df['Glucose'] != 0]
df = df[df['BMI'] != 0]
df.head()
```

```python
q1=df['Insulin'].quantile(0.25)
q3=df['Insulin'].quantile(0.75)
iqr = q3-q1
lw = q1-(1.5*iqr)
uw = q3-(1.5*iqr)
```

```python
outliers=df[(df['Insulin']>uw) | (df['Insulin']<lw)]
outliers.head()
```

```python
df1=df[(df['Insulin']<uw) & (df['Glucose']>lw)]
df1.head()
```

```python
q1=df1['Age'].quantile(0.25)
q3=df1['Age'].quantile(0.75)
iqr=q3-q1
lw = q1-(1.5*iqr)
uw = q3-(1.5*iqr)
```

```python
x=df.iloc[:,:5]
y=df['Outcome']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.24,random_state=0,stratify=y)
```

```python
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()

nb.fit(x_train,y_train)

y_pred=nb.predict(x_test)
y_pred
```

```python
from sklearn.metrics import ConfusionMatrixDisplay,confusion_matrix
con_mtx=confusion_matrix(y_test,y_pred)
con_mtx
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
score=accuracy_score(y_test,y_pred)
score
```

```python
from sklearn.metrics import precision_score,confusion_matrix
pre=precision_score(y_test,y_pred)
pre
```

```python
from sklearn.metrics import recall_score,confusion_matrix
rec=recall_score(y_test,y_pred)
rec
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
