```python
def conv_Mileage(row):
    row = row.lower().strip()
    if row.endswith('km/kg'):
        num = row.replace('km/kg','')
        return float(num)*1.4
    elif row.endswith('kmpl'):
         num = row.replace('kmpl','')
         return float(num)
        
df['Mileage'] = df['Mileage'].apply(conv_Mileage)
```

r2_score = 1 - ss_res/ss_tot
