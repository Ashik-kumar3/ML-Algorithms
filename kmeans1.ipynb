```python
import pandas as pd
```

```python
df = pd.read_csv("kmeansData.csv",header=None)
df.head()
```

```python
df.shape
```

```python
from sklearn.cluster import KMeans

Kmeans = KMeans(n_clusters=3,init="k-means++")

Kmeans.fit(df)

Kmeans.inertia_
```

'k-means++': selects initial cluster centroids using sampling based on an emperical probability of distribution of the points contribution to the overall inertia. This technique speeds up convergence.

```python
y_kmeans=Kmeans.predict(df)
y_kmeans
```

```python
from sklearn.cluster import KMeans
wcss = []

for k in range(1,10):
    kmeans1 = KMeans(n_clusters=k,init='k-means++',random_state=0)
    kmeans1.fit(df)
    wcss.append(kmeans1.inertia_)
```

```python
wcss
```

```python
import matplotlib.pyplot as plt

plt.plot(range(1,10),wcss,marker='*')

plt.xlabel('number of clusters')
plt.ylabel('inertia')
plt.show()
```

```python
Kmeans = KMeans(n_clusters=2,init='k-means++',random_state=0)
y_kmeans = Kmeans.fit_predict(df)
y_kmeans
```

```python
df['cluster']=y_kmeans
df
```
